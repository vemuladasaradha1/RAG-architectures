# 08 — Contextual RAG
**What:** retrieve chunks with surrounding context and metadata. **Why:** reduce ambiguity caused by isolated chunks. **When:** documents have sections, references or repeated terminology.

```python
from rag_architectures.core.pipeline import docs_from_pairs
from rag_architectures.architectures.all import ContextualRAG
docs=docs_from_pairs([('Context','A chunk can depend on its surrounding section context.'),('RAG','RAG retrieves supporting evidence.')])
rag=ContextualRAG(docs); query='Why does surrounding context matter?' 
```

```python
[(h.score,h.document.metadata,h.document.text) for h in rag.retrieve(query,3)]
```

```python
r=rag.run(query,3); print(r.answer); print(r.trace)
```
