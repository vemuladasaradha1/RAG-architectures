# 09 — Parent-Child RAG
**What:** index small child chunks but return their larger parent context. **Why:** precise matching plus coherent context. **When:** long structured documents.

```python
from rag_architectures.core.pipeline import docs_from_pairs
from rag_architectures.architectures.all import ParentChildRAG
docs=docs_from_pairs([('Parent','A parent document contains several related child passages.'),('Child','A child passage is a precise retrieval unit.')])
rag=ParentChildRAG(docs); query='What is a child passage used for?' 
```

```python
[(h.score,h.document.text) for h in rag.retrieve(query,3)]
```

```python
r=rag.run(query,3); print(r.answer); print(r.trace)
```
