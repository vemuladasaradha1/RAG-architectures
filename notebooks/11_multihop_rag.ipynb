# 11 — Multi-Hop RAG
**What:** retrieve in multiple steps. **Why:** answer questions requiring connected evidence. **When:** one retrieval pass cannot find all facts.

```python
from rag_architectures.core.pipeline import docs_from_pairs
from rag_architectures.architectures.all import MultiHopRAG
docs=docs_from_pairs([('Step1','Alice works at Acme.'),('Step2','Acme is headquartered in Paris.'),('Hop','Multi-hop retrieval follows intermediate evidence.')])
rag=MultiHopRAG(docs); query='Where is the company associated with Alice headquartered?' 
```

```python
[(h.score,h.document.text) for h in rag.retrieve(query,3)]
```

```python
r=rag.run(query,3); print(r.answer); print(r.trace)
```
