```python
from rag_architectures.core.pipeline import docs_from_pairs
from rag_architectures.architectures.all import HyDERAG
docs=docs_from_pairs([('HyDE','HyDE retrieves using a hypothetical document.'),('RAG','RAG uses external evidence.')])
rag=HyDERAG(docs); query='Explain hypothetical-document retrieval' 
```
