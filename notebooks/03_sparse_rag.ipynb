# 03 — Sparse / BM25 RAG
**What:** lexical retrieval. **Why:** excellent for exact terms. **When:** identifiers, names, rare words.

```python
from rag_architectures.core.pipeline import docs_from_pairs
from rag_architectures.architectures.all import SparseRAG
docs=docs_from_pairs([('RAG','Retrieval augmented generation.'),('BM25','BM25 ranks documents using lexical term statistics.')])
rag=SparseRAG(docs); query='BM25 lexical term statistics' 
```

```python
hits=rag.retrieve(query,3); [(h.score,h.document.text) for h in hits]
```

```python
r=rag.run(query,3); print(r.answer); print(r.trace)
```
