# Basic RAG

Baseline lifecycle: **ingest → parse → chunk → index → retrieve → optional transform/rerank/verify → generate → evaluate**.

This notebook makes every stage visible and keeps the example API-key free.

```python
from rag_architectures.core.pipeline import docs_from_pairs
from rag_architectures.architectures.all import BasicRAG
docs = docs_from_pairs([
    ('RAG', 'Retrieval augmented generation retrieves relevant evidence before an LLM generates an answer.'),
    ('BM25', 'BM25 is a sparse lexical retrieval method.'),
    ('Embeddings', 'Dense embeddings represent text as vectors for semantic retrieval.'),
])
rag = BasicRAG(docs)
query = 'What does retrieval augmented generation retrieve before generation?'
```

```python
print('INGEST:', len(rag.ingested_documents), 'source documents')
print('PARSE:', len(rag.parsed_documents), 'parsed documents')
print('CHUNK:', len(rag.chunks), 'retrieval chunks')
print('INDEX:', type(rag.store).__name__)
```

```python
result = rag.run(query, k=3)
print('RETRIEVE:')
for hit in result.retrieved:
    print(round(hit.score, 3), '->', hit.document.text)
```

```python
print('OPTIONAL TRANSFORM/RERANK/VERIFY:')
for key, value in result.trace['optional_stages'].items():
    print(key, ':', value)
```

```python
print('GENERATE:', result.answer)
print('EVALUATE:', result.trace['evaluation'])
assert [s['step'] for s in result.trace['pipeline']] == ['ingest','parse','chunk','index','retrieve','optional transform/rerank/verify','generate','evaluate']
```

### Takeaway
Basic RAG keeps the pipeline simple. Production systems replace the demo parser/index/LLM/evaluator with real components, but the lifecycle remains observable.
