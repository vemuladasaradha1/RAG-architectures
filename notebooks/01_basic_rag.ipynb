```python
from rag_architectures.core.pipeline import docs_from_pairs
from rag_architectures.architectures.all import BasicRAG
docs=docs_from_pairs([('RAG','RAG retrieves evidence before generation.'),('BM25','BM25 is lexical retrieval.'),('Graph','Graphs connect entities.')])
rag=BasicRAG(docs)
query='What does RAG retrieve before generation?' 
```

```python
hits=rag.retrieve(query,3)
[(h.score,h.document.text) for h in hits]
```

```python
result=rag.run(query,3)
print(result.answer)
print(result.trace)
```
