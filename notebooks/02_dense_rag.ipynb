# 02 — Dense / Semantic RAG
**What:** semantic similarity retrieval. **Why:** handles paraphrases. **When:** wording differs from source text.

```python
from rag_architectures.core.pipeline import docs_from_pairs
from rag_architectures.architectures.all import DenseRAG
docs=docs_from_pairs([('RAG','Retrieval augmented generation retrieves evidence before generation.'),('BM25','BM25 is lexical retrieval.')])
rag=DenseRAG(docs); query='How does RAG find supporting evidence?' 
```

```python
hits=rag.retrieve(query,3); [(h.score,h.document.text) for h in hits]
```

```python
r=rag.run(query,3); print(r.answer); print(r.trace)
```
