# 04 — Hybrid RAG
Dense + BM25 retrieval fusion using LangChain.

Pipeline: ingest → parse → chunk → index → retrieve → optional transform/rerank/verify → generate → evaluate.

```python
import langchain
from rag_architectures.architectures.all import HybridRAG
from rag_architectures.core.pipeline import ingest_html, parse_html, chunk_documents, build_index, langchain_generator
print('LangChain:',langchain.__version__)
raw_html=ingest_html('../data/sample/mtech_quantum_project.html')
parsed=parse_html(raw_html)
chunks=chunk_documents(parsed,700,100)
index=build_index(chunks)
print('PARSE',len(parsed),'CHUNK',len(chunks),'INDEX',type(index).__name__)
```

```python
query='What are the key implementation steps and expected outcomes in the M.Tech project?'
result=HybridRAG().run(raw_html,query)
print('Stages:',' → '.join(s['stage'] for s in result.trace.stages))
print('GENERATE:',langchain_generator().invoke({'question':query,'contexts':result.retrieved}))
print('EVALUATE:',result.evaluation)
assert result.answer
```
