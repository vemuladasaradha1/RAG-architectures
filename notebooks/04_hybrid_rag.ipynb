# 04 — Hybrid RAG
**What:** combine semantic and lexical retrieval. **Why:** better recall across paraphrases and exact terms. **When:** production general-purpose search.

```python
from rag_architectures.core.pipeline import docs_from_pairs
from rag_architectures.architectures.all import HybridRAG
docs=docs_from_pairs([('Semantic','Semantic retrieval finds related meaning.'),('BM25','BM25 finds exact lexical terms.'),('RAG','RAG uses retrieved evidence.')])
rag=HybridRAG(docs); query='RAG retrieval using lexical terms' 
```

```python
[(h.score,h.source,h.document.text) for h in rag.retrieve(query,3)]
```

```python
r=rag.run(query,3); print(r.answer); print(r.trace)
```

Production: use Qdrant hybrid queries or a vector DB plus BM25/OpenSearch, then tune fusion weights. No API key required.
